```python
categorical_columns = [
    "gender",
    "age_group",
    "appointment_type",
    "appointment_day",
    "appointment_time",
    "reminder_sent",
    "reminder_channel",
    "appointment_outcome"
]

for column in categorical_columns:
    print(f"\n{'=' * 50}")
    print(column.upper())
    print("=" * 50)
    print(df[column].value_counts(dropna=False))
```

```python
numeric_columns = [
    "age",
    "booking_lead_days",
    "previous_appointments",
    "previous_no_shows",
    "distance_to_clinic_km",
    "waiting_time_minutes"
]

df[numeric_columns].describe().T
```

```python
missing_summary = pd.DataFrame({
    "Missing Values": df.isna().sum(),
    "Missing Percentage": (
        df.isna().sum() / len(df) * 100
    ).round(2)
})

missing_summary = missing_summary[
    missing_summary["Missing Values"] > 0
].sort_values(
    "Missing Values",
    ascending=False
)

missing_summary
```

```python
pd.crosstab(
    df["reminder_sent"],
    df["reminder_channel"],
    dropna=False
)
```

```python
df["booking_date"] = pd.to_datetime(
    df["booking_date"],
    errors="coerce"
)

df["appointment_date"] = pd.to_datetime(
    df["appointment_date"],
    errors="coerce"
)
```

```python
print(df["booking_date"].dtype)
print(df["appointment_date"].dtype)
```

```python
print(
    "Booking Date Range:",
    df["booking_date"].min().date(),
    "to",
    df["booking_date"].max().date()
)

print(
    "Appointment Date Range:",
    df["appointment_date"].min().date(),
    "to",
    df["appointment_date"].max().date()
)
```

```python
invalid_booking_dates = (
    df["booking_date"] > df["appointment_date"]
).sum()

print(
    f"Bookings occurring after appointment date: "
    f"{invalid_booking_dates:,}"
)
```

```python
previous_no_show_distribution = (
    df["previous_no_shows"]
    .value_counts()
    .sort_index()
)

previous_no_show_distribution
```
